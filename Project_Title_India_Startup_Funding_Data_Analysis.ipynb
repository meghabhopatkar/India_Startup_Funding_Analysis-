# Total Funding Raised

Purpose

Calculate the total startup funding available in the dataset.

```python
import duckdb

# Register the pandas DataFrame so DuckDB can query it as a table named 'startup'
startup = df

# Run the SQL query
query = """
SELECT SUM(Funding_Amount) AS Total_Funding
FROM startup;
"""

duckdb.query(query).to_df()
```

Top 10 Industries by Funding

Purpose

Identify industries attracting the highest investments.

```python
import duckdb

# Run the SQL query using DuckDB
query = """
SELECT Industry,
SUM(Funding_Amount) AS Total_Funding
FROM startup
GROUP BY Industry
ORDER BY Total_Funding DESC
LIMIT 10;
"""

duckdb.query(query).to_df()
```

Top 10 Cities by Funding

Purpose

Discover major startup hubs in India.

```python
import duckdb

# Run the SQL query using DuckDB with corrected column name
query = """
SELECT City,
SUM(Funding_Amount) AS Total_Funding
FROM startup
GROUP BY City
ORDER BY Total_Funding DESC
LIMIT 10;
"""

duckdb.query(query).to_df()
```

Top Investors by Funding

Purpose

Find investors contributing the largest amount of capital.

```python
import duckdb

# Run the SQL query using DuckDB with corrected column name
query = """
SELECT Lead_Investor,
SUM(Funding_Amount) AS Total_Funding
FROM startup
GROUP BY Lead_Investor
ORDER BY Total_Funding DESC;
"""

duckdb.query(query).to_df()
```

Funding Stage Distribution

Purpose

Analyze which funding stage is most common.

```python
import duckdb

# Run the SQL query using DuckDB
query = """
SELECT Funding_Stage,
COUNT(*) AS Number_of_Startups
FROM startup
GROUP BY Funding_Stage
ORDER BY Number_of_Startups DESC;
"""

duckdb.query(query).to_df()
```

Year-wise Funding Trend

Purpose

Understand how startup funding has changed over time.

```python
import duckdb

# Run the SQL query using DuckDB with corrected column name
query = """
SELECT Funding_Year,
SUM(Funding_Amount) AS Total_Funding
FROM startup
GROUP BY Funding_Year
ORDER BY Funding_Year;
"""

duckdb.query(query).to_df()
```

Average Funding by Industry

Purpose

Determine which sectors receive the highest average investment.

```python
import duckdb

# Run the SQL query using DuckDB with the corrected column name
query = """
SELECT Industry,
AVG(Funding_Amount) AS Avg_Funding
FROM startup
GROUP BY Industry
ORDER BY Avg_Funding DESC;
"""

duckdb.query(query).to_df()
```

Top 5 Startups by Funding

Purpose

Identify startups that received the highest funding.

```python
import duckdb

# Run the SQL query using DuckDB with the corrected column name
query = """
SELECT Startup_Name,
Funding_Amount
FROM startup
ORDER BY Funding_Amount DESC
LIMIT 5;
"""

duckdb.query(query).to_df()
```

State-wise Funding Analysis

Purpose

Find states attracting the most startup investments.

```python
import duckdb

# Run the SQL query using DuckDB with corrected column name
query = """
SELECT State,
SUM(Funding_Amount) AS Total_Funding
FROM startup
GROUP BY State
ORDER BY Total_Funding DESC;
"""

duckdb.query(query).to_df()
```

Funding Stage vs Average Funding

Purpose

Compare investment amounts across different funding stages.

```python
import duckdb

# Run the SQL query using DuckDB with the corrected column name
query = """
SELECT Funding_Stage,
AVG(Funding_Amount) AS Avg_Funding
FROM startup
GROUP BY Funding_Stage
ORDER BY Avg_Funding DESC;
"""

duckdb.query(query).to_df()
```

Top Investor using RANK()

```python
import duckdb

# Run the SQL query using DuckDB with the corrected column name
query = """
SELECT Lead_Investor,
SUM(Funding_Amount) AS Total_Funding,
RANK() OVER(
ORDER BY SUM(Funding_Amount) DESC
) AS Investor_Rank
FROM startup
GROUP BY Lead_Investor;
"""

duckdb.query(query).to_df()
```

Industry Above Average Funding

```python
import duckdb

# Run the SQL query using DuckDB with corrected column name
query = """
SELECT Industry,
SUM(Funding_Amount) AS Total_Funding
FROM startup
GROUP BY Industry
HAVING SUM(Funding_Amount) >
(
SELECT AVG(Funding_Amount)
FROM startup
);
"""

duckdb.query(query).to_df()
```
